```python
import wfdb
import numpy as np
import matplotlib.pyplot as plt

patient_data = "Person_01/rec_5"
ecg = wfdb.rdrecord(patient_data)
annotations = wfdb.rdann(patient_data, 'atr')
voltages = ecg.p_signal
print(ecg.sig_name)
signal = voltages[:, 0]
fs = 500
```

```python
def highpass_filter(sig, cutoff, fs):
    import numpy as np
    
    sig = np.asarray(sig)
    y = np.zeros_like(sig)

    dt = 1/fs
    RC = 1/(2*np.pi*cutoff)
    alpha = RC/(RC + dt)

    for n in range(1, len(sig)):
        y[n] = alpha * (y[n-1] + sig[n] - sig[n-1])

    return y
    
hp = highpass_filter(signal, cutoff=5, fs=500)   

def lowpass_filter(sig, cutoff, fs):
    import numpy as np
    
    sig = np.asarray(sig)
    y = np.zeros_like(sig)

    dt = 1/fs
    RC = 1/(2*np.pi*cutoff)
    alpha = dt/(RC + dt)

    for n in range(1, len(sig)):
        y[n] = y[n-1] + alpha * (sig[n] - y[n-1])

    return y

signal = lowpass_filter(hp, cutoff=12, fs=500)
```

```python
alpha = 0.2
filtsig = np.zeros(len(signal))
filtsig[0] = signal[0]
for n in range(1, len(signal)):
    filtsig[n] = alpha*signal[n] + (1-alpha)*filtsig[n-1] #y(n) = ax(n) + (1-a)y(n-1)

plt.figure(figsize=(15,10))
plt.plot(signal, label='raw')
plt.plot(filtsig, label=f'EMA (alpha={alpha})')
plt.legend()
plt.show()
```

```python
squared = filtsig ** 2
#squared = squared / np.max(squared)
plt.figure(figsize=(15, 10))
plt.plot(filtsig, label='Filtered Signal')
plt.plot(squared, label='Squared')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(15, 10))
plt.plot(squared, label='Squared')
plt.legend()
plt.show()           
```

```python
def find_local_peaks(sig):
    peaks = []
    for i in range(1, len(sig)-1):
        if sig[i] > sig[i-1] and sig[i] > sig[i+1]:
            peaks.append(i)
    return np.array(peaks)

def threshold_peaks(sig, peaks, thresh_ratio=0.3):
    thresh = thresh_ratio * np.max(sig)
    return peaks[sig[peaks] > thresh]

def enforce_refractory(sig, peaks, fs, refractory=0.1):
    min_samples = int(refractory * fs)
    filtered = []

    last = -np.inf
    for p in peaks:
        if p - last > min_samples:
            filtered.append(p)
            last = p

    return np.array(filtered)
    
def detect_r_peaks(sig, fs):
    peaks = find_local_peaks(sig)
    peaks = threshold_peaks(sig, peaks, 0.3)
    peaks = enforce_refractory(sig, peaks, fs, 0.2)
    return peaks
    
r_peaks = detect_r_peaks(squared, fs)    
```

```python
plt.figure(figsize=(15,10))
plt.plot(signal)
plt.scatter(r_peaks, signal[r_peaks], color='red')
plt.show()
```

```python
def instantaneous_hr(r_peaks, fs):
    hr = []
    
    for i in range(1, len(r_peaks)):
        rr = (r_peaks[i] - r_peaks[i-1]) / fs
        hr.append(60 / rr)
        
    return np.array(hr)

hr = instantaneous_hr(r_peaks, fs)
for h in hr :
    print("HR is : ", h)
```
